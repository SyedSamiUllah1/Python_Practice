```python
import numpy as np
import matplotlib.pyplot as plt

xvalues = np.linspace(-np.pi, np.pi)

yvalues1 = np.sin(xvalues)
yvalues2 = np.cos(xvalues)

plt.plot(
    xvalues,
    yvalues1,
    lw=2,
    color='red',
    label='sin(x)'
)

plt.plot(
    xvalues,
    yvalues2,
    lw=2,
    color='blue',
    label='cos(x)'
)

plt.title('Trigonometric Functions')
plt.xlabel('x')
plt.ylabel('sin(x), cos(x)')

plt.axhline(0, lw=0.5, color='black')
plt.axvline(0, lw=0.5, color='black')

plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Initialize a square figure
plt.figure(figsize=(6, 6))

# Generate 100 evenly spaced numbers between -2 and 2
xvalues = np.linspace(-2, 2, 100)

# --- Subplot 1 (Top-Left) ---
plt.subplot(3, 3, 1)
yvalues = xvalues
plt.plot(xvalues, yvalues, color='blue')
plt.xlabel('$x$')
plt.ylabel('$x$')

# --- Subplot 2 (Top-Right) ---
plt.subplot(3, 3, 2)
yvalues = xvalues ** 2
plt.plot(xvalues, yvalues, color='green')
plt.xlabel('$x$')
plt.ylabel('$x^2$')

# --- Subplot 3 (Bottom-Left) ---
plt.subplot(3, 3, 3)
yvalues = xvalues ** 3
plt.plot(xvalues, yvalues, color='red')
plt.xlabel('$x$')
plt.ylabel('$x^3$')

# --- Subplot 4 (Bottom-Right) ---
plt.subplot(3, 3, 4)
yvalues = xvalues ** 4
plt.plot(xvalues, yvalues, color='black')
plt.xlabel('$x$')
plt.ylabel('$x^4$')  # Fixed label from $x^3$ to $x^4$ to match the function

#--- subplot 5 (Bottom centre)
plt.subplot(3,3,5)
yvalues=xvalues**5
plt.plot(xvalues,yvalues,color='black')
plt.xlabel('$x$')
plt.ylabel('$x^5$')

#--- subplot 6 (bottom right)
plt.subplot(3,3,6)
yvalues=xvalues**6
plt.plot(xvalues,yvalues,color='black')
plt.xlabel('$x$')
plt.ylabel('$x^6$')

# --- Global Layout Settings ---
plt.suptitle('Polynomial Functions')
plt.tight_layout()
plt.subplots_adjust(top=0.90)

# Display the plots
plt.show()
```
