Merging Joining

```python
employees = pd.DataFrame({
    'Employee_ID':[1,2,3,4],
    'Employee_Name':['Ali','Sara','Ahmed','Fatima'],
    'Department_ID':[101,102,101,103]
})

employees
```

```python
departments = pd.DataFrame({
    'Department_ID':[101,102,103],
    'Department_Name':['IT','HR','Sales']
})

departments
```

```python
pd.merge(employees, departments, on='Department_ID')
```

```python
pd.merge(
    employees,
    departments,
    how='left',
    on='Department_ID'
)
```

```python
pd.merge(
    employees,
    departments,
    how='right',
    on='Department_ID'
)
```

```python
pd.merge(
    employees,
    departments,
    how='outer',
    on='Department_ID'
)
```

```python
pd.merge(
    employees[['Employee_Name','Department_ID']],
    departments[['Department_ID','Department_Name']],
    on='Department_ID'
)
```
