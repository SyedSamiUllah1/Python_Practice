```python
np.random.seed(42)

df2 = pd.DataFrame({
    'Employee_ID': range(1001, 1031),
    'Employee_Name': [
        'Ali','Sara','Ahmed','Fatima','Bilal','Ayesha','Usman','Noor','Hamza','Zain',
        'Hina','Omar','Laiba','Saad','Iqra','Daniyal','Maham','Taha','Sana','Areeb',
        'Anas','Kiran','Abdullah','Maryam','Asad','Rabia','Huzaifa','Muneeb','Mehwish','Raza'
    ],
    'Department': np.random.choice(
        ['HR','IT','Finance','Marketing','Sales'],
        30
    ),
    'City': np.random.choice(
        ['Lahore','Karachi','Islamabad'],
        30
    ),
    'Gender': np.random.choice(
        ['Male','Female'],
        30
    ),
    'Experience': np.random.randint(1,11,30),
    'Salary': np.random.randint(40000,120001,30),
    'Bonus': np.random.randint(5000,25001,30),
    'Performance': np.random.choice(
        ['Excellent','Good','Average','Poor'],
        30
    )
})

df2
```

```python
gg=df2.groupby(['Department','Gender'])
```

```python
for name, group in gg:
    print(name)
    print("\n\n")
    print(group)
    print("\n\n")
```
