```python
df=pd.read_csv(r'dataset/Employees.csv')
```

```python
df['City']=df['City'].map({'Multan':'M'})
```

```python
df.sample(5)
```

```python
# for column, data in df.items():
#     print(column)
#     print(data)
for col, values in df.head(5).items():
    print(f"{col}: {values}")
```

```python
for index, row in df.head(5).iterrows():
    print(index)
    print(row)
```

```python
for index, row in df.head(5).iterrows():
    print(f"{index}{row['Employee_Name']}||{row['City']}")
```

```python
for row in df.head(5).itertuples():
    print(row)
```

```python
for index, row in df.iterrows():
    df.loc[index, 'Tax'] = row['Salary'] * 0.10
```
